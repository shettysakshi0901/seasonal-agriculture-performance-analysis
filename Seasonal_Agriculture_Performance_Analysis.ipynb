## 2. Problem Statement

Agricultural activities are influenced by seasonal variations in
environmental conditions, farming practices, resource availability,
and market conditions. As a result, agricultural performance may
differ from one season to another.

However, raw agricultural data does not clearly explain how
agricultural performance changes across seasons or what patterns
can be observed under different seasonal conditions.

Therefore, this project focuses on analyzing the given agricultural
dataset and investigating seasonal differences in agricultural
performance by identifying meaningful patterns, trends, relationships,
and variations within the available data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("Columns in the dataset:")

for i, column in enumerate(df.columns, 1):
    print(i, "-", column)
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

missing_summary[missing_summary["Missing Values"] > 0]
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing numerical values handled.")
```

```python
categorical_columns = df.select_dtypes(include="str").columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Missing categorical values handled.")
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)
```

```python
df.describe()
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=crop_yield.index,
    y=crop_yield.values
)

plt.title("Average Crop Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)

plt.show()
```

```python
crop_season_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

print("Average Crop Yield by Crop and Season:")
display(crop_season_yield)
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt=".2f"
)

plt.title("Average Crop Yield by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Rainfall_mm",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Relationship Between Rainfall and Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Soil_Moisture_pct",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Relationship Between Soil Moisture and Crop Yield")
plt.xlabel("Soil Moisture (%)")
plt.ylabel("Yield (Tonnes/Ha)")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Avg_Temperature_C",
    y="Yield_Tonnes_Ha",
    hue="Season"
)

plt.title("Relationship Between Temperature and Crop Yield")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Yield (Tonnes/Ha)")

plt.show()
```

```python
numeric_data = df.select_dtypes(include=np.number)

correlation = numeric_data.corr()

display(correlation)
```

```python
plt.figure(figsize=(14,10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Between Agricultural Variables")
plt.show()
```

```python
state_yield = df.groupby("State")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)

print("Average Yield by State:")
print(state_yield)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=state_yield.index,
    y=state_yield.values
)

plt.title("Average Crop Yield by State")
plt.xlabel("State")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)

plt.show()
```

```python
state_season_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="State",
    columns="Season",
    aggfunc="mean"
)

display(state_season_yield)
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="Season",
    y="Yield_Tonnes_Ha"
)

plt.title("Yield Variation and Outliers Across Seasons")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45)

plt.show()
```

```python
season_summary = df.groupby("Season").agg({
    "Rainfall_mm": "mean",
    "Soil_Moisture_pct": "mean",
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Total_Cost_INR": "mean",
    "Revenue_INR": "mean",
    "Profit_INR": "mean",
    "Water_Used_m3": "mean"
})

display(season_summary)
```

```python
season_summary.round(2)
```

```python
best_yield_season = season_summary["Yield_Tonnes_Ha"].idxmax()
worst_yield_season = season_summary["Yield_Tonnes_Ha"].idxmin()

best_profit_season = season_summary["Profit_INR"].idxmax()
worst_profit_season = season_summary["Profit_INR"].idxmin()

print("Highest average yield season:", best_yield_season)
print("Lowest average yield season:", worst_yield_season)

print("Highest average profit season:", best_profit_season)
print("Lowest average profit season:", worst_profit_season)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14,5))

sns.barplot(
    x=season_summary.index,
    y=season_summary["Yield_Tonnes_Ha"],
    ax=axes[0]
)

axes[0].set_title("Average Yield by Season")
axes[0].set_xlabel("Season")
axes[0].set_ylabel("Yield (Tonnes/Ha)")
axes[0].tick_params(axis="x", rotation=45)

sns.barplot(
    x=season_summary.index,
    y=season_summary["Profit_INR"],
    ax=axes[1]
)

axes[1].set_title("Average Profit by Season")
axes[1].set_xlabel("Season")
axes[1].set_ylabel("Profit (INR)")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

#  Key Findings and Insights

Based on the analysis performed on the agricultural dataset, the following key findings were observed:

### 1. Seasonal Yield Variation
Kharif recorded the highest average crop yield, followed by Rabi, while Zaid recorded the lowest average yield.

### 2. Seasonal Profit Variation
Kharif recorded the highest average profit. Rabi recorded positive profit, while Zaid showed negative average profit.

### 3. Environmental Conditions
Rainfall, soil moisture, and temperature varied across the different seasons. These variations provide useful context for understanding seasonal agricultural performance.

### 4. Crop Performance
Crop performance varied across different crops and seasons. The Crop × Season analysis helped identify differences in yield under different seasonal conditions.

### 5. Resource Usage
Water usage and irrigation practices differed across seasons, indicating differences in agricultural resource requirements.

### 6. Regional Variation
Agricultural yield varied across states, showing that agricultural performance is not uniform across geographical regions.

### 7. Relationships Between Variables
Correlation analysis and scatter plots were used to examine relationships between environmental conditions, resource usage, and agricultural yield.

### 8. Unusual Patterns
The boxplot analysis showed variation in yield across seasons and helped identify potential unusual observations and outliers.
